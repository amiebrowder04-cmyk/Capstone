## Merging DataFrames:

```python
# We are going to start with the HUD data and the nonprofit data so that after we combine them, I can aggregate by county before combining with the census data 
np_hud = np_clean.merge(
    cw_clean[["zip", "County_ID"]],
    left_on = "ZIP",
    right_on = "zip",
    how = "left"
)

#check
np_hud.head()
```

```python
# When merging both "zip" columns, we need to drop one
np_hud = np_hud.drop( columns = ["zip"])  

#check
np_hud.head()
```

```python
#creating a county-level non-profit count 
nonprofit_count = np_hud.groupby("County_ID")["EIN"].nunique().reset_index()

#check
nonprofit_count.head()
```

```python
nonprofit_count = nonprofit_count.rename(columns={"EIN": "Nonprofit_count"})

nonprofit_count.head()
```

```python
# adding in the nonprofit types to the nonprofit_count table 
np_hud["NTEE_Category"] = np_hud["NTEE_CD"].str[0]
#check
np_hud[["NTEE_CD", "NTEE_Category"]].head(20)
```

```python
# Now we are going to count each nonprofit category within each county 
ntee_counts = (
    np_hud.groupby(["County_ID", "NTEE_Category"])["EIN"]
    .nunique()
    .unstack(fill_value = 0)
    .reset_index()
)
#check
ntee_counts.head()
```

```python
# combining the nonprofit_count and the ntee_count tables 
nonprofit_count = nonprofit_count.merge(
    ntee_counts,
    on = "County_ID",
    how = "left"
)
#check

nonprofit_count.head()
```

```python
p_clean.head()
```

```python
# merging the new nonprofit_count table with the census table 
final_df = p_clean.merge(
    nonprofit_count,
    on = "County_ID",
    how = "left"
)
#check
final_df.head()
```

```python
# creating derived variables in order to make comparing nonprofits to the county population easier 

final_df["Total_Population"] = pd.to_numeric(
    final_df["Total_Population"],
    errors = "coerce"
)

final_df["Nonprofits_Per_1000"] = (
    final_df["Nonprofit_count"] / final_df["Total_Population"]
) * 1000

final_df["People_Per_Nonprofit"] = (
    final_df["Total_Population"] / final_df["Nonprofit_count"]
)

#check
final_df.head()
```

```python
#saving the final_df to a csv file to be opened and used later 
final_df.to_csv("final_combined_data.csv", index = False)
```
